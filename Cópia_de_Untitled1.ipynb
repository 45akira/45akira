```python
import matplotlib.pyplot as plt
import random

# Lista de nomes atualizada
nomes = ['Ariane', 'Nathália', 'Enzo', 'Conguito', 'Moisés']

# Cores para os setores da roleta
cores = ['red', 'blue', 'green', 'orange', 'purple']

# Criar gráfico de pizza (roleta)
fig, ax = plt.subplots()
ax.pie([1]*len(nomes), labels=nomes, colors=cores, startangle=90, counterclock=False)
ax.set_title('Roleta de Nomes 🎡')
plt.show()

# Escolher um nome aleatoriamente
escolhido = random.choice(nomes)
print(f'O nome sorteado foi: 🎉 {escolhido} 🎉')
```
